```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.svm import SVC, LinearSVC
from sklearn.multiclass import OneVsRestClassifier
from imblearn.pipeline import Pipeline
from imblearn.over_sampling import SMOTE
from sklearn.cluster import KMeans
from sklearn.metrics import calinski_harabasz_score
```

# 1

```python
data = pd.read_csv('../data/Frogs_MFCCs.csv', na_values="na")
```

```python
X = data[[c for c in data.columns if c.startswith("MFCCs")]]
y = data[["Family", "Genus", "Species"]]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

## B

### i

When using exact match, accuracy is scored 0 or 1 depending if there was at least 1 catagory wrong or all are correct. On the other hand, hamming score method takes a proportion, the proportion of catagories for a specific observation that are correct out of all catagories. Because exact match is much more strict than hamming, it will always be less than or equal to the hamming score.

### ii

```python
sigmas = np.linspace(0.1, 3, 10)
param_grid = {"estimator__C": np.logspace(-2, 4, 7), "estimator__gamma": 1 / (2 * sigmas**2)}

#family
g_fam = GridSearchCV(OneVsRestClassifier(SVC(kernel="rbf")), param_grid, cv=10, n_jobs=-1)
g_fam.fit(X_train, y_train["Family"])
print("Family:", g_fam.best_params_, g_fam.best_score_)

# genus
g_gen = GridSearchCV(OneVsRestClassifier(SVC(kernel="rbf")), param_grid, cv=10, n_jobs=-1)
g_gen.fit(X_train, y_train["Genus"])
print("Genus:", g_gen.best_params_, g_gen.best_score_)

# species
g_spe = GridSearchCV(OneVsRestClassifier(SVC(kernel="rbf")), param_grid, cv=10, n_jobs=-1)
g_spe.fit(X_train, y_train["Species"])
print("Species:", g_spe.best_params_, g_spe.best_score_)


pred = np.column_stack([g_fam.predict(X_test), g_gen.predict(X_test), g_spe.predict(X_test)])
true = y_test.values

print("Exact match:  ", np.mean(np.all(true == pred, axis=1)))
print("Hamming score:", np.mean(true == pred))
print("Hamming loss: ", 1 - np.mean(true == pred))
```

### iii

```python
param_grid = {"C": np.logspace(-3, 3, 13)}

# Family
g_fam = GridSearchCV(LinearSVC(penalty="l1", dual=False, max_iter=10000), param_grid, cv=10, n_jobs=-1, verbose=1)
g_fam.fit(X_train, y_train["Family"])
print("Family:", g_fam.best_params_, g_fam.best_score_)

# Genus
g_gen = GridSearchCV(LinearSVC(penalty="l1", dual=False, max_iter=10000), param_grid, cv=10, n_jobs=-1, verbose=1)
g_gen.fit(X_train, y_train["Genus"])
print("Genus:", g_gen.best_params_, g_gen.best_score_)

# Species
g_spe = GridSearchCV(LinearSVC(penalty="l1", dual=False, max_iter=10000), param_grid, cv=10, n_jobs=-1, verbose=1)
g_spe.fit(X_train, y_train["Species"])
print("Species:", g_spe.best_params_, g_spe.best_score_)

# Combine test predictions
pred = np.column_stack([g_fam.predict(X_test), g_gen.predict(X_test), g_spe.predict(X_test)])
true = y_test.values

print("Exact match:  ", np.mean(np.all(true == pred, axis=1)))
print("Hamming score:", np.mean(true == pred))
print("Hamming loss: ", 1 - np.mean(true == pred))
```

```python
param_grid = {"clf__C": np.logspace(-3, 3, 13)}

# Family
pipe_fam = Pipeline([("smote", SMOTE(random_state=42)), ("clf", LinearSVC(penalty="l1", dual=False, max_iter=10000))])
g_fam = GridSearchCV(pipe_fam, param_grid, cv=10, n_jobs=-1, verbose=1)
g_fam.fit(X_train, y_train["Family"])
print("Family:", g_fam.best_params_, g_fam.best_score_)

# Genus
pipe_gen = Pipeline([("smote", SMOTE(random_state=42)), ("clf", LinearSVC(penalty="l1", dual=False, max_iter=10000))])
g_gen = GridSearchCV(pipe_gen, param_grid, cv=10, n_jobs=-1, verbose=1)
g_gen.fit(X_train, y_train["Genus"])
print("Genus:", g_gen.best_params_, g_gen.best_score_)

# Species
pipe_spe = Pipeline([("smote", SMOTE(random_state=42)), ("clf", LinearSVC(penalty="l1", dual=False, max_iter=10000))])
g_spe = GridSearchCV(pipe_spe, param_grid, cv=10, n_jobs=-1, verbose=1)
g_spe.fit(X_train, y_train["Species"])
print("Species:", g_spe.best_params_, g_spe.best_score_)

# Combine test predictions
pred = np.column_stack([g_fam.predict(X_test), g_gen.predict(X_test), g_spe.predict(X_test)])
true = y_test.values

print("Exact match:  ", np.mean(np.all(true == pred, axis=1)))
print("Hamming score:", np.mean(true == pred))
print("Hamming loss: ", 1 - np.mean(true == pred))
```

When comparing using exact match, the gaussian kernals and one vs all classifier performed the best with a score of 0.989 compared to 0.91 with L1 penalized and 0.85 with smote. This is seen similarly in the hamming score, with scores of 0.992 for gaussian, 0.94 for L1 and 0.92 for smote. It is expected that each method has a higher score for the hamming vs exact match, but it is important to note theat gaussian/one vs all performed the best out of all the models used.

```python
hamming_distances = []
hamming_scores = []
hamming_losses = []
chosen_ks = []

for run in range(50): # perform the procedure 50 times
    best_k = None
    best_ch = -1
    best_cluster_ids = None

    # a)
    for k in range(2, 51):
        km = KMeans(n_clusters=k, n_init=3, random_state=run)
        cluster_ids = km.fit_predict(X)
        ch = calinski_harabasz_score(X, cluster_ids)

        if ch > best_ch:
            best_ch = ch
            best_k = k
            best_cluster_ids = cluster_ids

    #b)
    pred_family = np.empty(len(data), dtype=object)
    pred_genus = np.empty(len(data), dtype=object)
    pred_species = np.empty(len(data), dtype=object)

    for c in range(best_k):
        in_cluster = (best_cluster_ids == c)
        majority_family = data.loc[in_cluster, "Family"].value_counts().idxmax()
        majority_genus = data.loc[in_cluster, "Genus"].value_counts().idxmax()
        majority_species = data.loc[in_cluster, "Species"].value_counts().idxmax()
        pred_family[in_cluster] = majority_family
        pred_genus[in_cluster] = majority_genus
        pred_species[in_cluster] = majority_species

    # c)
    wrong_family = (pred_family != data["Family"].values)
    wrong_genus = (pred_genus != data["Genus"].values)
    wrong_species = (pred_species != data["Species"].values)

    wrong_per_row = wrong_family.astype(int) + wrong_genus.astype(int) + wrong_species.astype(int)
    distance = wrong_per_row.mean() 
    loss = distance / 3  
    score = 1 - loss

    hamming_distances.append(distance)
    hamming_losses.append(loss)
    hamming_scores.append(score)
    chosen_ks.append(best_k)

print("Hamming distance: mean", np.mean(hamming_distances), "std", np.std(hamming_distances))
print("Hamming score: mean", np.mean(hamming_scores), "std", np.std(hamming_scores))
print("Hamming loss: mean", np.mean(hamming_losses), "std", np.std(hamming_losses))
```
